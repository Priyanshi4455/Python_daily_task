```python
#Dictionary
data = {}
print(type(data))

data = dict()
print(type(data))
print(data)
```

```python
#Accessing Dictionary Values

#Using Key
data = {"name": "Rahul", "age": 26, "Att": True}
print(data)

data["name"] = "Priya"
print(data)
print(data["name"])
print(data["college"])
```

```python
#Using get()
data = {"name": "Priya", "age": 26, "Att": True}
print(data)

print(data.get("name"))
print(data.get("college"))
```

```python
#Predefined Dictionary Methods

#len()
data = {"name": "Priya", "age": 26, "Att": True}
print(len(data))
```

```python
#values()
data = {"name": "Priya", "age": 26, "Att": True}
print(data.values())
```

```python
#items()
data = {"name": "Priya", "age": 26, "Att": True}
print(data.items())
```

```python
#pop
data = {"name": "Priya", "age": 26, "Att": True}
data.pop("age")
print(data)
```

```python
#popitem()
data = {"name": "Priya", "age": 26, "Att": True}
print(data.popitem())
print(data)
```

```python
#clear()
data = {"name": "Priya", "age": 26, "Att": True}
data.clear()
print(data)
```

```python
#copy()
data = {"name": "Mohit", "age": 26, "Att": True}
newData = data.copy()
print(newData)

data["name"] = "Rohit"
print(newData)
print(data)
```

```python
#Membership Operator in Dictionary
data = {"name": "Mohit", "age": 26, "Att": True}
print("name" in data)
print("Mohit" in data.values())
```

```python
data = {"name": "Mohit", "age": 26, "Att": True}
print("name" not in data)
print("Mohit" not in data.values())
```
