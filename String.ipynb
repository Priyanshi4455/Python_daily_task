```python
#String
data = "Python Programming"
print(data[0])      #indexing
print(data[0:6])    #slicing
print(data[-1])     #negative indexing
```

```python
#Repitition Operator(*)
data = "Python"
print(data*3)
print(*[data]*3)    #unpacking
```

```python
#String Function

#len()
data = "Python Programming"
print(len(data))
```

```python
#upper()
data = "Python Programming"
print(data.upper())
```

```python
#lower()
data = "Python Programming"
print(data.lower())
```

```python
#strip
data = "   Python Programming   "
print(data.strip())
```

```python
#lstrip()
data = "   Python Programming"
print(data.lstrip())
```

```python
#rstrip()
data = "Python Programming   "
print(data.rstrip())
```

```python
#replace()
data = "Python Programmming"
newData = data.replace("Python" , "Java")
print(data)
print(newData)
```

```python
#find()
data = "Python Programmming"
print(data.find('P'))
print(data.find('Programming'))
print(data.find('z'))
```

```python
#inde()
data = "Python Programmming"
print(data.index('P'))
print(data.index('z'))
```

```python
#count()
data = "Python Programmming"
print(data.count('P'))
print(data.count('p'))
```

```python
#startswith()
email = "admin@gmail.com"
print(email.startswith('admin'))

data = "Python Programmming"
print(data.startswith('Python'))
```

```python
#endswith()
email = "admin@gmail.com"
print(email.endswith('com'))
print('@' in email)    #Membership Operator
```

```python
#split()
data = "I love Python Programming"
newData.split()
print(data)
print(newData)
```

```python
data = ['I', 'Love', 'Python', 'Programming']
newData = " ".join(data)
print(newData)
```

```python
#isalpha()
data = "Invertis"
print(data.isalpha())
```
