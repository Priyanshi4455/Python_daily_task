# Section 1: Basic Python Programs

## Q1. Print all even numbers from 1 to 20

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
print()
```

## Q2. Print all odd numbers from 1 to 20

```python
for i in range(1, 21):
    if i % 2 != 0:
        print(i, end=" ")
print()
```

## Q3. Print multiples of 5 from 5 to 50

```python
for i in range(5, 51, 5):
    print(i, end=" ")
print()
```

## Q4. Find the sum of numbers from 1 to 10

```python
total = 0
for i in range(1, 11):
    total += i
print("Sum =", total)
```

## Q5. Find the sum of even numbers from 1 to 100

```python
total = 0
for i in range(2, 101, 2):
    total += i
print("Sum of even numbers =", total)
```

## Q6. Find the sum of odd numbers from 1 to 100

```python
total = 0
for i in range(1, 101, 2):
    total += i
print("Sum of odd numbers =", total)
```

## Q7. Print the multiplication table of 7

```python
for i in range(1, 11):
    print(f"7 x {i} = {7 * i}")
```

## Q8. Take a number from the user and print its multiplication table

```python
n = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{n} x {i} = {n * i}")
```

## Q9. Print numbers from 1 to 100 divisible by both 3 and 5

```python
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print(i, end=" ")
print()
```

## Q10. Count even and odd numbers between 1 and 50

```python
even = 0
odd = 0
for i in range(1, 51):
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("Even numbers =", even)
print("Odd numbers =", odd)
```

## Q11. Print list elements in reverse order

```python
data = [22, 33, 12, 32, 45, 13]
for i in range(len(data) - 1, -1, -1):
    print(data[i], end=" ")
print()
```

## Q12. Find the largest number in a list using a for loop

```python
data = [22, 55, 12, 89, 34, 67]
largest = data[0]
for num in data:
    if num > largest:
        largest = num
print("Largest number =", largest)
```

## Q13. Find the smallest number in a list

```python
data = [22, 55, 12, 89, 34, 67]
smallest = data[0]
for num in data:
    if num < smallest:
        smallest = num
print("Smallest number =", smallest)
```

## Q14. Count vowels in a string

```python
text = input("Enter a string: ")
count = 0
for ch in text.lower():
    if ch in "aeiou":
        count += 1
print("Number of vowels =", count)
```

## Q15. Count the number of spaces in a sentence

```python
text = "Python is easy to learn"
count = 0
for ch in text:
    if ch == " ":
        count += 1
print("Number of spaces =", count)
```
