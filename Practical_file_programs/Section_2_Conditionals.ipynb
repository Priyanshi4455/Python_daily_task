## Q1. ATM Withdrawal

```python
pin = 1234
balance = 10000
entered_pin = int(input("Enter PIN: "))
amount = int(input("Enter withdrawal amount: "))

if entered_pin == pin:
    if amount % 100 == 0:
        if amount <= balance:
            balance -= amount
            print("Withdrawal Successful")
            print("Remaining Balance =", balance)
        else:
            print("Insufficient Balance")
    else:
        print("Amount must be a multiple of 100")
else:
    print("Invalid PIN")
```

## Q2. Student Result System

```python
m1 = int(input("Enter marks of subject 1: "))
m2 = int(input("Enter marks of subject 2: "))
m3 = int(input("Enter marks of subject 3: "))

if m1 >= 40 and m2 >= 40 and m3 >= 40:
    percentage = (m1 + m2 + m3) / 3
    print("Percentage =", percentage)
    if percentage >= 75:
        print("Distinction")
    elif percentage >= 60:
        print("First Division")
    elif percentage >= 50:
        print("Second Division")
    else:
        print("Pass")
else:
    print("Fail")
```

## Q3. Bank Loan Eligibility

```python
age = 25
salary = 30000
credit_score = 720

if age >= 21:
    if salary >= 25000:
        if credit_score >= 750:
            print("Loan Approved")
        elif credit_score >= 650:
            print("Loan Approved with Higher Interest")
        else:
            print("Loan Rejected")
    else:
        print("Salary Too Low")
else:
    print("Age Criteria Not Met")
```

## Q4. E-Commerce Discount

```python
amount = 6000
premium = True

if amount >= 5000:
    discount = 20 if premium else 10
else:
    discount = 5 if premium else 0

final_amount = amount - amount * discount / 100
print("Discount =", discount, "%")
print("Final Amount =", final_amount)
```

## Q5. Login + Role-Based Access

```python
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        role = input("Enter role: ")
        if role == "admin":
            print("Full Access")
        elif role == "teacher":
            print("Teacher Dashboard")
        elif role == "student":
            print("Student Dashboard")
        else:
            print("Invalid Role")
    else:
        print("Wrong Password")
else:
    print("Invalid Username")
```

## Q6. Movie Ticket Booking

```python
seats_available = True
age = 10
ticket_price = 200

if seats_available:
    if age < 5:
        price = 0
    elif age <= 12:
        price = ticket_price * 0.50
    elif age > 60:
        price = ticket_price * 0.70
    else:
        price = ticket_price
    print("Ticket Price =", price)
else:
    print("House Full")
```

## Q7. Electricity Bill

```python
units = 250

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
elif units <= 300:
    bill = units * 10
else:
    bill = units * 15

if bill > 3000:
    bill *= 1.10

print("Electricity Bill =", bill)
```

## Q8. Employee Bonus

```python
experience = 3
salary = 60000
performance = 85

if experience >= 1 and salary >= 50000:
    if performance >= 90:
        bonus_percent = 20
    elif performance >= 75:
        bonus_percent = 10
    else:
        bonus_percent = 5
elif salary < 50000:
    bonus_percent = 15 if performance >= 90 else 5
else:
    bonus_percent = 0

bonus = salary * bonus_percent / 100
print("Bonus Percentage =", bonus_percent)
print("Bonus Amount =", bonus)
```

## Q9. Driving License

```python
age = 20
learner_license = True
test_score = 75

if age >= 18:
    if learner_license:
        if test_score >= 60:
            print("Eligible for Driving License")
        else:
            print("Driving Test Failed")
    else:
        print("Learner License Required")
else:
    print("Underage")
```

## Q10. ATM - Difficult Challenge

```python
pin = 1234
entered_pin = 1234
active = True
balance = 30000
amount = 25000

if entered_pin != pin:
    print("Invalid PIN")
elif not active:
    print("Account Inactive")
elif amount <= 0:
    print("Invalid Amount")
elif amount % 100 != 0:
    print("Amount must be a multiple of 100")
elif balance - amount < 500:
    print("Minimum balance of 500 required")
else:
    if amount > 20000:
        print("Extra Verification Required")
    balance -= amount
    print("Withdrawal Successful")
    print("Remaining Balance =", balance)
```
