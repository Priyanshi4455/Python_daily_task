```python
#Set
data = {45,22,55,84,65,87,95,}
print(type(data))
print(data)
```

```python
numbers = {11,24,56,24,11,11,97,89}
print(numbers)
```

```python
#Set Methods

#add()
s = {1, 2}
s.add(3)
print(s)
```

```python
#update()
s = {1, 2}
s.update([3, 4], {5, 6})
print(s)
```

```python
#remove()
s = {11, 25, 78, 98, 86, 94}
s.remove(11)
print(s)
```

```python
#discard()
s = {24,56,45,67,89,78,95}
s.discard(45)
print(s)
```

```python
#pop()
s = {24,56,45,67,89,78,95}
s.pop()
print(s)
```

```python
#clear()
s = {11, 25, 78, 98, 86, 94}
s.clear()
print(s)
```

```python
#Looping through Set

s = {22,45,78,89}
for item in s:
    print(item)
```

```python
#Set Operations

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

#Union:All unique elements from both
print(A | B)
print(A.union(B))
```

```python
#Intersection:Common elements
print(A & B)
print(A.intersection(B))
```

```python
#Difference:Elements in A but not in B
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A - B)
print(A.difference(B))
```

```python
#Symmetric Difference:Elements in A or B, but not Both
print(A ^ B)
print(A.symmetric_difference(B))
```

```python
#Membership Operator
s = {22,45,78,89}
print(22 in s)
print(57 not in  s)
```
