# Python OOP – 1. Abstraction

## 1. What is abstraction?

Abstraction means **showing essential features and hiding unnecessary implementation details**.

For example, when you use an ATM, you choose withdraw or check balance. You do not need to understand the internal banking network or cash-dispensing mechanism.

In Python, abstraction is commonly implemented using:
- `ABC` (Abstract Base Class)
- `@abstractmethod` (a method that subclasses must implement)

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
```

## 2. Implementing an abstract class

`Payment` defines *what* every payment method must do, but not *how* it does it.

```python
class UpiPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")

class CardPayment(Payment):
    def pay(self, amount):
        print(f"Paid ₹{amount} using a debit/credit card")

upi = UpiPayment()
card = CardPayment()

upi.pay(1500)
card.pay(2500)
```

## 3. Why can't we create an object of an abstract class?

An abstract class is an incomplete blueprint. Python prevents direct instantiation while abstract methods remain unimplemented.

```python
try:
    payment = Payment()
except TypeError as error:
    print(error)
```

## 4. Another example: Shape

Different shapes calculate area differently. The abstract class defines a common interface.

```python
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

shapes = [Circle(5), Rectangle(4, 6)]
for shape in shapes:
    print(f"{type(shape).__name__} area: {shape.area():.2f}")
```

## 5. Abstract class with a regular method

An abstract class can also contain fully implemented methods that child classes inherit.

```python
class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

    def power_source(self):
        return "Electricity"

class Fan(Appliance):
    def turn_on(self):
        return "Fan is rotating"

fan = Fan()
print(fan.turn_on())
print(fan.power_source())
```

### Quick revision

- Abstraction focuses on **what an object does**, not the internal details of how it does it.
- Import `ABC` and `abstractmethod` from Python's `abc` module.
- A concrete child class must implement all inherited abstract methods before it can be instantiated.
