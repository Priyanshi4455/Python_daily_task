```python
#Tuple Packing
studentData = "Aman", "MCA", 78.65, 
print(type(studentData))

#Tuple Unpacking
name, course, avgMarks = studentData
print(name)
print(avgMarks)
```

```python
record = "Amit", "MCA", 23, True    #Tuple Packing
print(record[0])
print(record[1])
print(record)

name,course,age,att=record = record     #Tuple Unpacking

print(name)

#Tuple Unpacking Using *
n,*ab=record
#Note: The * operator collects the remaining elements into a list.
print(n)
print(ab)
```
