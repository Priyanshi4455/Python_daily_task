# Python OOP – 4. Polymorphism

## 1. What is polymorphism?

Polymorphism means “many forms”. The same method call or interface can produce different behavior depending on the object.

```python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Cow:
    def speak(self):
        return "Moo!"

animals = [Dog(), Cat(), Cow()]
for animal in animals:
    print(animal.speak())
```

## 2. Polymorphism through method overriding

A child class changes the inherited method's behavior.

```python
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

for shape in [Square(4), Rectangle(4, 6)]:
    print(shape.area())
```

## 3. Duck typing

Python often cares about whether an object supports the required behavior, rather than its exact class.

```python
class PDFDocument:
    def open(self):
        print("Opening PDF document")

class WordDocument:
    def open(self):
        print("Opening Word document")

def open_document(document):
    document.open()

open_document(PDFDocument())
open_document(WordDocument())
```

## 4. Operator overloading

Special methods such as `__add__` let custom objects work with operators.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)
print(p1 + p2)
```

**Remember:** polymorphism lets different object types be used through a shared interface.
