# Python OOP – 2. Encapsulation

## 1. What is encapsulation?

Encapsulation means bundling data and the methods that operate on that data inside a class, while controlling how that data is accessed or changed.

Python naming conventions:
- `name` – public attribute
- `_name` – protected-by-convention attribute
- `__name` – name-mangled attribute (not truly private)

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._course = "BCA"
        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Marks must be between 0 and 100.")

student = Student("Aarav", 85)
print(student.name)
print(student._course)  # Accessible, but intended for internal use
print(student.get_marks())
student.set_marks(92)
print(student.get_marks())
```

## 2. Name mangling

A double-underscore attribute is internally renamed by Python to reduce accidental access and name clashes. It is not a security mechanism.

```python
print(student.__dict__)
# The attribute is stored under a mangled name such as _Student__marks.
print(student._Student__marks)  # Demonstration only; normally use the public method.
```

## 3. Using properties

`@property` lets you access a method like an attribute while validating updates through a setter.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance cannot be negative.")
        self.__balance = amount

account = BankAccount("Meera", 5000)
print(account.balance)
account.balance = 6500
print(account.balance)

try:
    account.balance = -100
except ValueError as error:
    print(error)
```

**Remember:** encapsulation helps protect an object's valid state and keeps related data and behavior together.
