```python
data = [22, 45, 65, 89, 97, 57, 78]

largest = data[0]
i = 0

while i < len(data):
    if data[i] > largest:
        largest = data[i]
    i = i + 1

print("Largest Number:", largest)
```

```python
name = "Rahul"
vowels = "aeiouAEIOU"
count = 0
i = 0

while i < len(name):
    if name[i] in vowels:
        count += 1
    i = i + 1

print("Number of vowels:", count)
```
