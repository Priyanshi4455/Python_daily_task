# Python OOP – 3. Inheritance

## 1. What is inheritance?

Inheritance allows a child class to reuse and extend attributes and methods from a parent class.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

class Student(Person):
    def study(self, subject):
        print(f"{self.name} is studying {subject}.")

student = Student("Riya", 20)
student.introduce()
student.study("Python")
```

## 2. Using `super()`

`super()` calls a method from the parent class, commonly its constructor.

```python
class Person:
    def __init__(self, name):
        self.name = name

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def details(self):
        print(f"{self.name} teaches {self.subject}.")

teacher = Teacher("Karan", "Computer Science")
teacher.details()
```

## 3. Types of inheritance

- **Single:** one child inherits from one parent.
- **Multilevel:** a class inherits from a child class, forming a chain.
- **Multiple:** one child inherits from more than one parent.
- **Hierarchical:** multiple child classes inherit from one parent.
- **Hybrid:** a combination of inheritance patterns.

```python
# Multilevel inheritance
class LivingThing:
    def breathe(self):
        print("Breathing")

class Animal(LivingThing):
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()
dog.breathe()
dog.eat()
dog.bark()
```

```python
# Multiple inheritance
class Camera:
    def take_photo(self):
        print("Photo captured")

class Phone:
    def make_call(self):
        print("Calling...")

class SmartPhone(Camera, Phone):
    pass

device = SmartPhone()
device.take_photo()
device.make_call()
```

## 4. Method overriding

A child class can provide its own version of a parent method.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Cat(Animal):
    def sound(self):
        print("Meow")

Cat().sound()
```

**Remember:** inheritance supports code reuse, but use it when there is a genuine “is-a” relationship.
