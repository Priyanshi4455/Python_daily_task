```python
data=[1,2,3,4,5,6,7,8,9]
empty=[]
for i in data:
    x=i**2
    empty.append(x)
print(empty)
```

```python
data=[1,2,3,4,5,6,7,8,9]
empty=[x**2 for x in data]
print(empty)
```

```python
data=[25,65,78,83,98,88,89]
even=[x for x in data if x%2==0]
print(even)
```

```python
data=[45,79,56,34,89,94,75]
result=["Pass" if x>=60 else "Fail" for x in data]
print(result)
```

```python
names=["Amit", "Priya", "Raman", "Sneha", "Tanya"]
result=[n.upper() for n in names]
print(result)
```

```python
names=["Amit", "Priya", "Raman", "Sneha", "Tanya"]
result=[n.lower() for n in names]
print(result)
```

```python
names=["Amit  ", "  Priya ", "  Raman", "Sneha   ", "    Tanya ", "   Sakshi", "   Priyanka "]
cleaned_Data=[x.strip() for x in names]
print(names)
print(cleaned_Data)
```
