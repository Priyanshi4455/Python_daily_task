# Student Grade Calculator

```python
subjects = ["Python", "Mathematics", "Computer Fundamentals", "English", "Practical"]
marks = []

for subject in subjects:
    while True:
        try:
            score = float(input(f"Enter marks for {subject} (0-100): "))
            if 0 <= score <= 100:
                marks.append(score)
                break
            print("Marks must be between 0 and 100.")
        except ValueError:
            print("Please enter a valid number.")

total = sum(marks)
percentage = total / len(subjects)

if any(score < 33 for score in marks):
    result = "Needs improvement (one or more subjects below 33)"
elif percentage >= 90:
    result = "Grade A+"
elif percentage >= 80:
    result = "Grade A"
elif percentage >= 70:
    result = "Grade B"
elif percentage >= 60:
    result = "Grade C"
elif percentage >= 50:
    result = "Grade D"
else:
    result = "Grade E"

print("\\n--- REPORT CARD ---")
for subject, score in zip(subjects, marks):
    print(f"{subject:24} {score:6.1f}/100")
print(f"Total: {total:.1f}/{len(subjects)*100}")
print(f"Percentage: {percentage:.2f}%")
print(f"Result: {result}")
```
